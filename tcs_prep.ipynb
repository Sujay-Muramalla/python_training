```python
def zipped_strings(str1,str2):
    dict1 = dict(zip(str1,str2))
    #print (dict1)
    dict2 = {k.upper():v for k,v in dict1.items()}
    return dict2

str1=['Sujay',"john","adam"]
str2=[1,2,3]

all_caps = zipped_strings(str1,str2)

print (type(all_caps))


for k,v in all_caps.items():
    print (f"{k}: {v}")
```
